#### 1.1.1 기본 폴더의 위치와 역할

**목적:** 프로젝트 안에서 학습 이미지·학습 라벨·테스트 이미지가 어디에 저장되어 있는지 확인한다.


**시각화 사전 설정:** 실행 환경의 한글 글꼴을 자동으로 선택해 이후 matplotlib 그래프의 한글이 깨지지 않도록 한다.


```python
from pathlib import Path, PurePosixPath
from collections import Counter, defaultdict
import json
from IPython.display import Markdown, display
import matplotlib.pyplot as plt
from matplotlib import font_manager

# 저장소 루트 또는 notebooks 폴더에서 실행할 수 있도록 경로를 찾는다.
cwd = Path.cwd().resolve()
project_root = next(
    (p for p in [cwd, *cwd.parents] if (p / "data/raw").is_dir()),
    None,
)
if project_root is None:
    raise FileNotFoundError("프로젝트의 data/raw 폴더를 찾지 못했습니다.")

data_dir = project_root / "data/raw"
folder_roles = {
    "train_images": "학습 이미지",
    "train_annotations": "학습 정답 라벨",
    "test_images": "테스트 이미지",
}

def show_table(headers, rows):
    # 파일명·약품명에 포함될 수 있는 표 구분 문자와 줄바꿈을 처리한다.
    def text(value):
        return str(value).replace("|", "&#124;").replace("\n", " ")
    lines = [
        "| " + " | ".join(map(text, headers)) + " |",
        "| " + " | ".join(["---"] * len(headers)) + " |",
    ]
    lines.extend("| " + " | ".join(map(text, row)) + " |" for row in rows)
    display(Markdown("\n".join(lines)))

# 시각화 전에 현재 실행 환경에서 사용할 수 있는 한글 글꼴을 자동 선택한다.
korean_font_candidates = [
    "NanumGothic",
    "Noto Sans CJK KR",
    "Noto Sans KR",
    "Malgun Gothic",
    "AppleGothic",
]
available_font_names = {font.name for font in font_manager.fontManager.ttflist}
selected_korean_font = next(
    (name for name in korean_font_candidates if name in available_font_names),
    None,
)
plt.rcParams["axes.unicode_minus"] = False
if selected_korean_font:
    plt.rcParams["font.family"] = selected_korean_font
    display(Markdown(f"**시각화 한글 글꼴:** `{selected_korean_font}` 사용"))
else:
    display(Markdown(
        "⚠️ **시각화 한글 글꼴을 찾지 못했습니다.** "
        "현재 환경에 한글 글꼴을 설치한 뒤 커널을 다시 시작해야 한글이 정상 표시됩니다."
    ))

missing_folders = [
    name for name in folder_roles if not (data_dir / name).is_dir()
]
if missing_folders:
    raise FileNotFoundError(f"필수 폴더가 없습니다: {missing_folders}")

show_table(
    ["역할", "프로젝트 기준 경로", "존재 여부"],
    [
        [role, f"data/raw/{name}", "확인"]
        for name, role in folder_roles.items()
    ],
)
raw_subfolders = sorted(
    p.name for p in data_dir.iterdir()
    if p.is_dir() and not p.name.startswith(".")
)
display(Markdown("**data/raw의 하위 폴더:** " + ", ".join(raw_subfolders)))
```

**결과 설명:** `data/raw` 아래에 `train_images`, `train_annotations`, `test_images`가 있다. 별도의 검증 이미지 폴더는 확인되지 않았다.


#### 1.1.2 라벨 폴더의 하위 구조

**목적:** _json 항목의 실제 종류와 내부 계층을 확인해 JSON 파일의 탐색 방법을 정한다.

```python
from pathlib import Path
from itertools import islice

annotation_dir = data_dir / "train_annotations"

for item in islice(annotation_dir.iterdir(), 3):
    print(f"\n{item.name}")
    print("종류:", "폴더" if item.is_dir() else "파일")

    if item.is_dir():
        for child in islice(item.iterdir(), 3):
            print("  └", child.name)
```

**결과 설명:** 출력한 `_json` 항목은 JSON 파일이 아니라 폴더였고, 그 안에 `K-` 코드별 하위 폴더와 JSON 파일이 저장되어 있다. 따라서 하위 폴더까지 탐색해야 라벨 파일을 찾을 수 있다.


### 1.2 파일 형식·개수·용량

**목적:** 전체 하위 파일의 확장자·개수·용량을 확인해 학습 데이터 규모와 저장 공간을 파악한다.

```python
files_by_folder = {}
inventory_rows = []
zero_byte_paths = []
total_bytes = 0

for folder_name in folder_roles:
    folder = data_dir / folder_name
    files = sorted(
        p for p in folder.rglob("*")
        if p.is_file()
        and not any(
            part.startswith(".") for part in p.relative_to(folder).parts
        )
    )
    files_by_folder[folder_name] = files
    extension_counts = Counter(
        p.suffix.lower() or "(확장자 없음)" for p in files
    )
    sizes = {p: p.stat().st_size for p in files}
    folder_bytes = sum(sizes.values())
    total_bytes += folder_bytes
    empty_files = [p for p, size in sizes.items() if size == 0]
    zero_byte_paths.extend(empty_files)
    inventory_rows.append([
        folder_name,
        len(files),
        ", ".join(f"{ext}: {count}" for ext, count in sorted(extension_counts.items())),
        f"{folder_bytes / 1024**2:,.2f}",
        len(empty_files),
    ])

show_table(
    ["폴더", "파일 수", "확장자별 개수", "용량(MiB)", "0바이트 파일 수"],
    inventory_rows,
)
display(Markdown(
    f"**전체 원본 파일 용량:** {total_bytes / 1024**3:,.3f} GiB "
    "(MiB와 GiB는 각각 1024의 거듭제곱 기준)"
))

image_suffixes = {".png", ".jpg", ".jpeg", ".bmp", ".webp", ".tif", ".tiff"}
train_image_paths = [
    p for p in files_by_folder["train_images"]
    if p.suffix.lower() in image_suffixes
]
test_image_paths = [
    p for p in files_by_folder["test_images"]
    if p.suffix.lower() in image_suffixes
]
json_paths = [
    p for p in files_by_folder["train_annotations"]
    if p.suffix.lower() == ".json"
]
```

**결과 설명:** 학습 이미지는 PNG 232개, 테스트 이미지는 PNG 842개, 학습 라벨은 JSON 763개다. 전체 원본 용량은 약 1.793 GiB이며 0바이트 파일은 없었다. 0바이트가 아니라는 사실만으로 이미지가 정상적으로 열리는지는 판단할 수 없다.

## 2. JSON 전체 구조 요약

**목적:** JSON의 최상위 구조와 전체 파일의 구성을 확인하고,
파일 수·이미지 정보 수·객체 정답 수·클래스 정의 수를 구분한다.

JSON 객체는 Python에서 `dict`로 읽으며 `key: value` 쌍으로 구성된다.
`images`·`annotations`·`categories`의 값은 `list`이고, 각 목록의 항목은 다시 `dict`이다.
따라서 `images.id`와 `annotations.id`는 서로 다른 구조에 속한 key이다.

### 2.1 샘플 JSON의 최상위 구조

**목적:** JSON 한 파일의 최상위 key와 각 key에 저장된 자료형을 확인한다.


```python
import json
from pathlib import Path

annotation_dir = data_dir / "train_annotations"
json_path = next(annotation_dir.rglob("*.json"), None)

if json_path is None:
    print("JSON 파일을 찾지 못했어요.")
else:
    print("확인할 파일:", json_path.relative_to(annotation_dir))

    with json_path.open(encoding="utf-8-sig") as f:
        label = json.load(f)

    print("최상위 자료형:", type(label).__name__)

    if isinstance(label, dict):
        for key, value in label.items():
            print(f"- {key}: {type(value).__name__}")
    elif isinstance(label, list):
        print("항목 수:", len(label))
        if label and isinstance(label[0], dict):
            print("첫 항목의 키:", list(label[0].keys()))
```

**결과 설명:** 샘플 JSON의 최상위는 dict이며 `images`, `type`, `annotations`, `categories` 네 key가 보였다. `images`·`annotations`·`categories`는 list, `type`은 str로 읽혔다.


### 2.2 샘플의 항목 개수와 유형

**목적:** 샘플 하나의 이미지·객체 정답·클래스 정의 항목 수를 확인하고 전체 집계와 구분한다.

```python
print("이미지 정보 개수:", len(label["images"]))
print("라벨 개수:", len(label["annotations"]))
print("클래스 정보 개수:", len(label["categories"]))
print("type의 실제 값:", label["type"])
```

**결과 설명:** 출력한 샘플 한 파일에는 이미지 정보·annotation·category가 각각 1개 있고, `type`의 값은 `instances`였다. 이 값은 샘플 1개의 결과이므로 전체 파일 집계와 구분한다.


### 2.3 전체 JSON의 구성과 집계 단위

**목적:** 샘플에서 본 구조가 전체 JSON에도 적용되는지 확인하고 읽기 실패와 구조 차이를 점검한다.

```python
image_rows = []
annotation_rows = []
category_rows = []
json_summary = []
json_errors = []
structure_errors = []

for json_path in json_paths:
    relative_path = str(json_path.relative_to(data_dir))
    try:
        with json_path.open(encoding="utf-8-sig") as f:
            document = json.load(f)
    except (OSError, UnicodeError, json.JSONDecodeError) as error:
        json_errors.append([relative_path, str(error)])
        continue

    if not isinstance(document, dict):
        structure_errors.append([relative_path, "JSON 최상위가 dict가 아님"])
        continue

    blocks = {}
    for key in ["images", "annotations", "categories"]:
        value = document.get(key)
        if not isinstance(value, list):
            structure_errors.append([relative_path, f"{key}가 list가 아님"])
            value = []
        if any(not isinstance(item, dict) for item in value):
            structure_errors.append([relative_path, f"{key}에 dict 외 항목이 있음"])
        blocks[key] = [item for item in value if isinstance(item, dict)]

    json_summary.append({
        "json_path": relative_path,
        "images": len(blocks["images"]),
        "annotations": len(blocks["annotations"]),
        "categories": len(blocks["categories"]),
    })
    for image in blocks["images"]:
        file_name = image.get("file_name")
        image_name = (
            PurePosixPath(file_name.replace("\\", "/")).name
            if isinstance(file_name, str) and file_name else None
        )
        image_rows.append({
            "json_path": relative_path,
            "image_name": image_name,
            "metadata": image,
        })
    annotation_rows.extend(
        {"json_path": relative_path, **item}
        for item in blocks["annotations"]
    )
    category_rows.extend(
        {"json_path": relative_path, **item}
        for item in blocks["categories"]
    )

show_table(
    ["집계 항목", "개수"],
    [
        ["JSON 파일", len(json_paths)],
        ["최상위 구조를 확인한 JSON", len(json_summary)],
        ["JSON 읽기 실패", len(json_errors)],
        ["예상 구조와 다른 항목", len(structure_errors)],
        ["images 항목 합계", len(image_rows)],
        ["annotations 원시 항목 합계", len(annotation_rows)],
        ["categories 항목 합계", len(category_rows)],
    ],
)
json_shapes = Counter(
    (row["images"], row["annotations"], row["categories"])
    for row in json_summary
)
show_table(
    ["JSON당 images", "JSON당 annotations", "JSON당 categories", "해당 JSON 수"],
    [[*shape, count] for shape, count in sorted(json_shapes.items())],
)
if json_errors or structure_errors:
    show_table(["파일", "확인 필요 사항"], (json_errors + structure_errors)[:10])
```

**결과 설명:** 전체 JSON 763개에서 key 구성은 파일 간 동일했다. 각 파일에는 이미지 정보·annotation·category가 각각 1개씩 있어, 763건의 annotation은 763개의 고유 이미지라는 뜻이 아니다.

```python
# Kaggle JSON 구조표의 '설명' 열 문구를 기재한다. 표 안의 줄바꿈은 공백으로 표시한다.
# 출처: https://www.kaggle.com/competitions/ai15-level1-project/data
# 해석·주의 사항은 정의와 구분하여 마크다운에 기록한다.
field_definitions = {
    "images": {
        "id": "약제 이미지식별자",
        "file_name": "약제 이미지파일명",
        "width": "약제 이미지너비",
        "height": "약제 이미지높이",
        "drug_N": "알약코드",
        "drug_S": "알약상태",
        "back_color": "촬영배경",
        "drug_dir": "알약방향",
        "light_color": "촬영조명",
        "camera_la": "카메라위도",
        "camera_lo": "카메라경도",
        "size": "이미지 사이즈",
        "dl_idx": "알약 식별자",
        "dl_mapping_code": "제품코드",
        "dl_name": "제품명",
        "dl_name_en": "제품명(영어)",
        "img_key": "이미지 링크",
        "dl_material": "성분명",
        "dl_material_en": "성분명(영어)",
        "dl_custom_shape": "제조 모양",
        "dl_company": "제조사명",
        "dl_company_en": "제조사명(영어)",
        "di_company_mf": "위탁제조사명",
        "di_company_mf_en": "위탁제조사명(영어)",
        "item_seq": "품목기준코드",
        "di_item_permit_date": "허가일자",
        "di_class_no": "약품 분류",
        "di_etc_otc_code": "전문의약품/일반의약품",
        "di_edi_code": "EDI 코드",
        "chart": "알약 제형",
        "drug_shape": "알약 모양",
        "thick": "알약 두께",
        "leng_long": "알약 장축",
        "leng_short": "알약 단축",
        "print_front": "식별문자_앞",
        "print_back": "식별문자_뒤",
        "color_class1": "색상 1",
        "color_class2": "색상 2",
        "line_front": "알약 앞면분할선 여부",
        "line_back": "알약 뒷면분할선 여부",
        "img_regist_ts": "약학정보원 이미지 생성일",
        "form_code_name": "정제 분류명",
        "mark_code_front_anal": "알약 앞면마크 형태",
        "mark_code_back_anal": "알약 뒷면마크 형태",
        "mark_code_front_img": "알약 앞면마크 이미지",
        "mark_code_back_img": "알약 뒷면마크 이미지",
        "mark_code_front": "알약 앞면마크 코드",
        "mark_code_back": "알약 뒷면마크 코드",
        "change_date": "변경일자"
    },
    "annotations": {
        "area": "바운딩박스 면적",
        "iscrowd": "평가 분류",
        "bbox": "bbox 좌표",
        "category_id": "category 아이디",
        "ignore": "무시 여부",
        "segmentation": "라벨링 바운딩 박스",
        "id": "어노테이션 아이디",
        "image_id": "이미지 아이디"
    },
    "categories": {
        "supercategory": "슈퍼 카테고리",
        "id": "카테고리 아이디",
        "name": "카테고리 명"
    }
}

# 파일 하나의 값이 아니라 전체 JSON의 값을 모은다.
field_records = {section: [] for section in field_definitions}
field_paths = sorted((data_dir / "train_annotations").rglob("*.json"))
field_read_errors = []
field_structure_errors = []
for field_path in field_paths:
    try:
        with field_path.open(encoding="utf-8-sig") as field_file:
            field_document = json.load(field_file)
    except (OSError, UnicodeError, json.JSONDecodeError) as error:
        field_read_errors.append([str(field_path), str(error)])
        continue
    if not isinstance(field_document, dict):
        field_structure_errors.append([str(field_path), "최상위가 dict가 아님"])
        continue
    for section in field_definitions:
        items = field_document.get(section)
        if not isinstance(items, list):
            field_structure_errors.append([str(field_path), section, "list가 아님"])
            continue
        for item in items:
            if isinstance(item, dict):
                field_records[section].append(item)
            else:
                field_structure_errors.append([str(field_path), section, "dict가 아닌 항목"])

if field_read_errors or field_structure_errors:
    raise ValueError(
        f"전체 필드 집계를 완료하지 못했습니다. 읽기 오류: {field_read_errors[:3]}, "
        f"구조 오류: {field_structure_errors[:3]}"
    )

def field_type_summary(values):
    types = {type(value).__name__ for value in values}
    if "list" in types:
        elements = [
            element for value in values if isinstance(value, list)
            for element in value
        ]
        element_types = {type(element).__name__ for element in elements}
        if element_types:
            types.remove("list")
            types.add("list[" + ", ".join(sorted(element_types)) + "]")
    return ", ".join(sorted(types)) if types else "미존재"

def field_value_summary(section, key, records):
    values = [record[key] for record in records if key in record]
    null_count = sum(value is None for value in values)
    empty_count = sum(isinstance(value, str) and value == "" for value in values)
    missing_count = len(records) - len(values)
    filled = [
        value for value in values
        if value is not None and not (isinstance(value, str) and value == "")
    ]
    parts = []
    if filled and all(
        isinstance(value, (int, float)) and not isinstance(value, bool)
        for value in filled
    ):
        unique = sorted(set(filled))
        if len(unique) == 1:
            parts.append(f"{unique[0]}로 동일")
        elif len(unique) <= 10:
            parts.append(", ".join(str(value) for value in unique))
            parts.append(f"고유 값 {len(unique)}개")
        else:
            parts.append(f"관찰 범위: {min(unique)}~{max(unique)}")
            parts.append(f"고유 값 {len(unique)}개")
        if key in {"id", "image_id", "category_id", "item_seq"}:
            parts.append("식별자 번호 기준")
    elif filled and all(isinstance(value, list) for value in filled):
        lengths = sorted({len(value) for value in filled})
        tokens = list(dict.fromkeys(
            json.dumps(value, ensure_ascii=False) for value in filled
        ))
        parts.append(f"{tokens[0]}로 동일" if len(tokens) == 1 else f"예: {tokens[0]}")
        parts.append("길이 " + ", ".join(map(str, lengths)))
        # 숫자 목록은 위치별 범위만 기록한다. 좌표 의미를 추정해서 이름 붙이지 않는다.
        if (
            len(lengths) == 1 and 0 < lengths[0] <= 8
            and all(
                isinstance(element, (int, float)) and not isinstance(element, bool)
                for value in filled for element in value
            )
        ):
            ranges = []
            for index in range(lengths[0]):
                column = [value[index] for value in filled]
                ranges.append(f"{min(column)}~{max(column)}")
            parts.append("원소별 관찰 범위: [" + ", ".join(ranges) + "]")
    elif filled:
        tokens = list(dict.fromkeys(
            json.dumps(value, ensure_ascii=False, sort_keys=True) for value in filled
        ))
        if len(tokens) == 1:
            parts.append(f"{tokens[0]}로 동일")
        elif len(tokens) <= 10:
            parts.append(", ".join(tokens))
        else:
            parts.append(f"예: {tokens[0]}")
        if len(tokens) > 1:
            parts.append(f"고유 값 {len(tokens)}개")
    if empty_count:
        parts.append(f'빈 문자열 "" {empty_count}건')
    if null_count:
        parts.append(f"null {null_count}건")
    if missing_count:
        parts.append(f"key 누락 {missing_count}건")
    return "; ".join(parts) if parts else "관찰 값 없음"

field_tables = {}
field_keyset_counts = {}
for section, definitions in field_definitions.items():
    records = field_records[section]
    observed_keys = {key for record in records for key in record}
    # 정의표의 항목 외에 실제 데이터에서 발견된 key도 생략하지 않는다.
    ordered_keys = (
        [key for key in definitions if key in observed_keys]
        + sorted(observed_keys - set(definitions))
    )
    field_tables[section] = [
        [
            key,
            definitions.get(key, "—"),
            field_value_summary(section, key, records),
            field_type_summary([record[key] for record in records if key in record]),
        ]
        for key in ordered_keys
    ]
    field_keyset_counts[section] = len({
        tuple(sorted(record)) for record in records
    })


# 각 필드의 정의와 실제 값은 한 용도에만 배치한다.
field_groups = {
    "image": {
        "images": ["id", "file_name", "imgfile", "width", "height", "size"],
        "annotations": ["image_id"],
    },
    "drug": {
        "images": [
            "drug_N", "dl_idx", "dl_mapping_code", "dl_name", "dl_name_en",
            "img_key", "dl_material", "dl_material_en", "dl_company", "dl_company_en",
            "di_company_mf", "di_company_mf_en", "item_seq", "di_item_permit_date",
            "di_class_no", "di_etc_otc_code", "di_edi_code", "img_regist_ts", "change_date",
        ],
        "categories": ["id", "name", "supercategory"],
        "annotations": ["category_id"],
    },
    "appearance": {
        "images": [
            "chart", "dl_custom_shape", "drug_shape", "form_code_name",
            "thick", "leng_long", "leng_short", "print_front", "print_back",
            "color_class1", "color_class2", "line_front", "line_back",
            "mark_code_front_anal", "mark_code_back_anal",
            "mark_code_front_img", "mark_code_back_img",
            "mark_code_front", "mark_code_back",
        ],
    },
    "capture": {
        "images": ["drug_S", "back_color", "light_color", "drug_dir", "camera_la", "camera_lo"],
    },
    "object": {
        "annotations": ["id", "bbox", "area", "iscrowd", "ignore", "segmentation"],
    },
}
field_rows_by_key = {
    f"{section}.{row[0]}": row
    for section, rows in field_tables.items() for row in rows
}
grouped_keys = [
    f"{section}.{key}"
    for blocks in field_groups.values()
    for section, keys in blocks.items() for key in keys
]
if len(grouped_keys) != len(set(grouped_keys)) or set(grouped_keys) != set(field_rows_by_key):
    raise ValueError("필드 분류에 누락 또는 중복이 있습니다. 실제 key와 용도별 분류를 확인하세요.")


def show_field_group(group):
    rows = []
    for section, keys in field_groups[group].items():
        for key in keys:
            qualified_key = f"{section}.{key}"
            rows.append([qualified_key, *field_rows_by_key[qualified_key][1:]])
    show_table(["key", "정의", "실제 값·관찰 범위", "자료형"], rows)


show_table(
    ["구조", "딕셔너리 항목 수", "전체 key 수", "관찰된 key 구성 종류 수"],
    [
        [section, len(field_records[section]), len(field_tables[section]), field_keyset_counts[section]]
        for section in field_definitions
    ],
)
```

### 3.2 이미지 크기 필드의 제공 범위와 일관성

**목적:** 같은 이미지가 여러 JSON에 기록될 때 크기 정보가 빠지거나 달라지는지 확인한다.

```python
metadata_fields = {
    "width": "기록된 이미지 너비",
    "height": "기록된 이미지 높이",
    "back_color": "배경색",
    "light_color": "조명",
    "drug_dir": "알약 앞뒷면",
    "camera_la": "카메라 관련 필드",
    "camera_lo": "카메라 관련 필드",
    "size": "size 필드(의미·단위 추가 확인)",
}
metadata_summary = []
metadata_details = {}

for field, description in metadata_fields.items():
    filled_count = 0
    value_tokens = set()
    values_by_image = defaultdict(set)
    for row in image_rows:
        value = row["metadata"].get(field)
        if value is None or (isinstance(value, str) and not value.strip()):
            continue
        filled_count += 1
        # 값의 자료형을 유지하며 비교한다.
        token = json.dumps(value, ensure_ascii=False, sort_keys=True)
        value_tokens.add(token)
        if row["image_name"]:
            values_by_image[row["image_name"]].add(token)

    value_examples = [str(json.loads(token)) for token in sorted(value_tokens)]
    conflict_count = sum(len(values) > 1 for values in values_by_image.values())
    metadata_details[field] = {
        "filled_records": filled_count,
        "unique_values": len(value_tokens),
        "images_with_value": len(values_by_image),
        "images_with_conflicts": conflict_count,
    }
    metadata_summary.append([
        field,
        description,
        f"{filled_count}/{len(image_rows)}",
        len(image_rows) - filled_count,
        len(values_by_image),
        len(value_tokens),
        ", ".join(value_examples[:5]),
        conflict_count,
    ])


def show_metadata_consistency(fields):
    show_table(
        ["key", "값 있는 images 항목", "결측 항목", "값 있는 고유 이미지", "동일 이미지 내 값 불일치 수"],
        [
            [f"images.{row[0]}", row[2], row[3], row[4], row[7]]
            for row in metadata_summary if row[0] in fields
        ],
    )


show_metadata_consistency({"width", "height", "size"})
```

**결과 설명:** `width`, `height`, `size`는 images 항목 763개 모두에 값이 있었고, 고유 이미지 232개 안에서 값 불일치는 없었다.


### 3.3 실제 이미지 파일과 JSON의 참조 관계

**목적:** 파일명 기준 참조 누락과 중복 파일명을 확인하고 이미지 한 장에 연결된 JSON 수를 파악한다.

### 3.4 JSON 내부의 이미지 ID 연결과 전체 ID 범위

**목적:** 각 객체 정답이 같은 JSON의 이미지 정보를 가리키는지 확인하고,
이미지 ID와 파일명의 대응을 근거로 여러 JSON을 함께 읽을 기준을 정한다.

**Kaggle 데이터 가공 기준:** `image_id`는 각 이미지의 파일명(`file_name`)을 기준으로 중복 없이 고유하게 부여된다. Kaggle 설명의 순차 부여는 전체 데이터 기준이다.

- 내부 연결: `annotations.image_id`와 **같은 JSON**의 `images.id`를 비교한다.
- 전체 대응: 하나의 이미지 ID가 여러 파일명에 쓰이는지, 하나의 파일명에 여러 ID가 기록되는지 확인한다.
- 같은 이미지가 여러 JSON에 반복 기록되는 것은 서로 다른 이미지에 ID가 재사용되는 경우와 구분한다.

**집계 단위:** 객체 연결은 annotation 항목 수, ID 재사용은 고유 ID 수,
파일명의 ID 불일치는 고유 파일명 수로 센다. ID는 정수 여부를 확인하며 원본 값을 변환하지 않는다.

**해석 기준:** 학습 데이터에서 파일명과 ID가 일대일로 대응하는지 확인한다. 학습 데이터는 전체 데이터의 일부이므로 ID 번호가 연속하지 않는다는 사실만으로 순차 부여 규칙의 오류를 뜻하지 않는다. 전체 데이터에서의 순번은 현재 학습 라벨만으로 검증하지 않는다.

```python
def valid_connection_id(value):
    return isinstance(value, int) and not isinstance(value, bool)


# JSON 경로를 포함해 조회하므로 다른 파일의 같은 번호와 연결되지 않는다.
images_by_json_id = defaultdict(list)
image_files_by_id = defaultdict(set)
image_ids_by_file = defaultdict(set)
invalid_image_id_rows = []
for row in image_rows:
    image_id = row["metadata"].get("id")
    if not valid_connection_id(image_id):
        invalid_image_id_rows.append(row)
        continue
    images_by_json_id[(row["json_path"], image_id)].append(row)
    if row["image_name"]:
        image_files_by_id[image_id].add(row["image_name"])
        image_ids_by_file[row["image_name"]].add(image_id)

image_link_checks = []
for annotation in annotation_rows:
    image_id = annotation.get("image_id")
    matches = (
        images_by_json_id.get((annotation["json_path"], image_id), [])
        if valid_connection_id(image_id) else []
    )
    if not valid_connection_id(image_id):
        status = "image_id 누락·형식 확인"
    elif not matches:
        status = "같은 JSON에서 이미지 ID를 찾지 못함"
    elif len(matches) > 1:
        status = "같은 JSON에서 이미지 ID가 여러 항목과 연결됨"
    else:
        status = "일치"
    image_link_checks.append({
        "json_path": annotation["json_path"],
        "annotation": annotation,
        "image_id": image_id,
        "image": matches[0] if status == "일치" else None,
        "status": status,
    })

image_link_status_counts = Counter(row["status"] for row in image_link_checks)
image_id_file_conflicts = {
    image_id: files for image_id, files in image_files_by_id.items() if len(files) > 1
}
image_file_id_conflicts = {
    name: ids for name, ids in image_ids_by_file.items() if len(ids) > 1
}
show_table(
    ["검사 항목", "결과"],
    [
        ["검사한 객체 정답 수", len(image_link_checks)],
        ["같은 JSON의 이미지와 연결된 객체 정답 수", image_link_status_counts["일치"]],
        ["id가 없거나 정수가 아닌 images 항목 수", len(invalid_image_id_rows)],
        ["image_id가 없거나 정수가 아닌 객체 정답 수", image_link_status_counts["image_id 누락·형식 확인"]],
        ["같은 JSON에서 이미지 ID를 찾지 못한 객체 정답 수", image_link_status_counts["같은 JSON에서 이미지 ID를 찾지 못함"]],
        ["같은 JSON에서 이미지 ID가 여러 항목과 연결된 객체 정답 수", image_link_status_counts["같은 JSON에서 이미지 ID가 여러 항목과 연결됨"]],
        ["고유 이미지 ID 수", len(image_files_by_id)],
        ["고유 이미지 파일명 수", len(image_ids_by_file)],
        ["서로 다른 파일명에 재사용된 이미지 ID 수", len(image_id_file_conflicts)],
        ["여러 이미지 ID가 기록된 파일명 수", len(image_file_id_conflicts)],
    ],
)

# 확인 대상은 종류별 최대 5건을 보여주고, 위 요약표에 전체 건수를 남긴다.
for status, count in sorted(image_link_status_counts.items()):
    if status == "일치":
        continue
    display(Markdown(f"**{status}: 전체 {count}건, 최대 5건 표시**"))
    show_table(
        ["JSON 경로(data/raw 기준)", "annotation ID", "image_id"],
        [
            [row["json_path"], row["annotation"].get("id"), row["image_id"]]
            for row in image_link_checks if row["status"] == status
        ][:5],
    )
if invalid_image_id_rows:
    show_table(
        ["id 형식 확인이 필요한 JSON(최대 5건)", "images.id"],
        [[row["json_path"], row["metadata"].get("id")] for row in invalid_image_id_rows[:5]],
    )
if image_id_file_conflicts:
    show_table(
        ["재사용된 이미지 ID(최대 5개)", "연결된 서로 다른 파일명"],
        [[key, " / ".join(sorted(values))] for key, values in sorted(image_id_file_conflicts.items())[:5]],
    )
if image_file_id_conflicts:
    show_table(
        ["여러 ID가 기록된 파일명(최대 5개)", "이미지 ID 목록"],
        [[key, sorted(values)] for key, values in sorted(image_file_id_conflicts.items())[:5]],
    )
```

**결과 설명:** annotation 763개 모두 같은 JSON의 `images.id`와 연결됐다. 고유 이미지 ID와 파일명은 각각 232개였고, 파일명과 ID는 일대일로 대응했다. ID 재사용이나 하나의 파일명에 여러 ID가 연결된 경우는 없었다. 학습 데이터에서 관찰된 ID는 14~1499 범위이며 연속하지 않는다. Kaggle 기준의 전체 데이터 순차 부여 여부는 학습 데이터만으로 판정하지 않는다.


### 3.5 이미지별 실제 연결 예시

**목적:** 한 이미지의 정보가 여러 JSON에 반복되고 각 객체 정답이 해당 이미지 ID로 연결되는 모습을 확인한다.

연결된 JSON 수가 2개·3개·4개인 이미지에서 각각 한 장을 선택한다.
각 그룹에서 파일명 정렬상 첫 이미지를 사용하므로 실행할 때마다 같은 예시를 볼 수 있다.
아래 표에는 선택한 이미지에 연결된 모든 annotation을 표시하며, 전체 연결 검사는 앞의 요약표 기준이다.

```python
image_link_samples = []
for target_json_count in [2, 3, 4]:
    candidates = sorted(
        name for name, paths in jsons_by_image.items()
        if len(paths) == target_json_count and name in train_names
    )
    if not candidates:
        display(Markdown(f"연결 JSON이 {target_json_count}개인 학습 이미지가 없습니다."))
        continue
    selected_name = candidates[0]
    selected_paths = jsons_by_image[selected_name]
    selected_links = [
        row for row in image_link_checks if row["json_path"] in selected_paths
    ]
    image_link_samples.append({
        "image_name": selected_name,
        "json_count": target_json_count,
        "annotation_count": len(selected_links),
    })
    display(Markdown(
        f"**연결 JSON {target_json_count}개인 이미지**\n\n`{selected_name}`"
    ))
    show_table(
        ["JSON 경로(data/raw 기준)", "images.id", "annotations.image_id", "같은 JSON 내 연결"],
        [
            [
                row["json_path"],
                row["image"]["metadata"].get("id") if row["image"] else "연결 미확정",
                row["image_id"], row["status"],
            ]
            for row in sorted(selected_links, key=lambda row: row["json_path"])
        ],
    )
```

### 3.6 파일명 코드와 JSON 이미지 코드 비교

**목적:** 파일명에서 읽은 약품 코드 후보와 연결된 JSON의 `images.drug_N`을 비교해, JSON에 나타나지 않는 코드가 있는 이미지 수와 코드 수를 집계한다.

파일명에 적힌 코드 의미가 공식적으로 모두 확인된 것은 아니므로, 차이는 **라벨 누락 확정이 아닌 누락 의심 후보**로 해석한다. 실제 이미지에서 해당 약품이 보이는지 확인한 뒤 누락 여부를 판단한다.


```python
import re

# 파일명 앞부분의 K-숫자 코드 목록과 연결 JSON의 images.drug_N을 비교한다.
drug_codes_by_image = defaultdict(set)
for row in image_rows:
    image_name = row.get("image_name")
    drug_code = row["metadata"].get("drug_N")
    if image_name and isinstance(drug_code, str) and drug_code.strip():
        drug_codes_by_image[image_name].add(drug_code.strip())

filename_json_code_checks = []
unparsed_filename_rows = []
for image_path in train_image_paths:
    image_name = image_path.name
    filename_prefix = image_path.stem.split("_", 1)[0]
    if not re.fullmatch(r"K(?:-\d{6})+", filename_prefix):
        unparsed_filename_rows.append(image_name)
        continue

    filename_codes = {
        f"K-{code}" for code in filename_prefix.split("-")[1:]
    }
    json_codes = drug_codes_by_image.get(image_name, set())
    missing_codes = filename_codes - json_codes
    extra_codes = json_codes - filename_codes
    filename_json_code_checks.append({
        "image_name": image_name,
        "filename_codes": filename_codes,
        "json_codes": json_codes,
        "missing_codes": missing_codes,
        "extra_codes": extra_codes,
        "json_count": len(jsons_by_image.get(image_name, set())),
    })

suspected_missing_rows = [
    row for row in filename_json_code_checks if row["missing_codes"]
]
extra_json_code_rows = [
    row for row in filename_json_code_checks if row["extra_codes"]
]
filename_json_code_count_distribution = Counter(
    (len(row["filename_codes"]), len(row["json_codes"]))
    for row in filename_json_code_checks
)

show_table(
    ["확인 항목", "결과"],
    [
        ["파일명 코드 추출에 성공한 학습 이미지", len(filename_json_code_checks)],
        ["파일명 코드가 JSON에도 모두 나타나는 이미지", sum(not row["missing_codes"] for row in filename_json_code_checks)],
        ["누락 의심 코드가 하나 이상 있는 이미지", len(suspected_missing_rows)],
        ["누락 의심 코드 수", sum(len(row["missing_codes"]) for row in suspected_missing_rows)],
        ["파일명에는 없고 JSON에만 있는 코드 수", sum(len(row["extra_codes"]) for row in extra_json_code_rows)],
        ["파일명 규칙을 읽지 못한 이미지", len(unparsed_filename_rows)],
    ],
)

show_table(
    ["파일명 코드 수", "JSON drug_N 코드 수", "이미지 수"],
    [
        [filename_count, json_count, count]
        for (filename_count, json_count), count
        in sorted(filename_json_code_count_distribution.items())
    ],
)

if suspected_missing_rows:
    show_table(
        ["이미지 파일명", "파일명 코드 후보", "JSON drug_N 코드", "누락 의심 코드", "연결 JSON 수"],
        [
            [
                row["image_name"],
                ", ".join(sorted(row["filename_codes"])),
                ", ".join(sorted(row["json_codes"])),
                ", ".join(sorted(row["missing_codes"])),
                row["json_count"],
            ]
            for row in sorted(suspected_missing_rows, key=lambda item: item["image_name"])
        ],
    )

if extra_json_code_rows:
    show_table(
        ["이미지 파일명", "파일명 코드 후보", "JSON에만 있는 코드"],
        [
            [row["image_name"], ", ".join(sorted(row["filename_codes"])), ", ".join(sorted(row["extra_codes"]))]
            for row in sorted(extra_json_code_rows, key=lambda item: item["image_name"])
        ],
    )

if unparsed_filename_rows:
    show_table(
        ["파일명 규칙 확인이 필요한 이미지"],
        [[name] for name in unparsed_filename_rows[:10]],
    )
```

### 3.7 대표 이미지·bbox·연결 라벨 시각화

**목적:** 파일명 코드와 JSON `drug_N`이 일치하는 일반 학습 이미지 중 객체 정답이 3개인 사례와 4개인 사례를 각각 골라 원본 이미지, bbox, 연결된 JSON 라벨을 함께 확인한다. 여러 JSON에 나뉘어 저장된 정답이 한 이미지에 모이는 모습을 비교한다.

각 사례는 bbox 네 값이 양수이며 기록된 이미지 크기 안에 들어오는 이미지 중 파일명순 첫 번째를 사용한다. bbox는 Kaggle 설명의 `[x, y, width, height]` 순서로 그린다. **초록색 bbox는 이 표본의 현재 확인 기준(파일명·JSON 코드 일치, bbox 형식과 이미지 경계 통과)을 통과했다는 뜻**이며, 전체 bbox 품질과 경계 오류는 7번에서 따로 검증한다.

```python
from matplotlib.patches import Rectangle
from PIL import Image

train_image_by_name = {path.name: path for path in train_image_paths}
filename_check_by_name = {row["image_name"]: row for row in filename_json_code_checks}
category_name_by_json_id = {
    (row["json_path"], row.get("id")): row.get("name")
    for row in category_rows
}

def bbox_fits_recorded_image(link):
    bbox = link["annotation"].get("bbox")
    metadata = link["image"]["metadata"]
    width, height = metadata.get("width"), metadata.get("height")
    if (
        not isinstance(bbox, list) or len(bbox) != 4
        or not all(isinstance(value, (int, float)) and not isinstance(value, bool) for value in bbox)
        or not isinstance(width, (int, float)) or not isinstance(height, (int, float))
    ):
        return False
    x, y, box_width, box_height = bbox
    return (
        x >= 0 and y >= 0 and box_width > 0 and box_height > 0
        and x + box_width <= width and y + box_height <= height
    )

def show_representative_image(target_count):
    for image_name in sorted(train_names):
        code_check = filename_check_by_name.get(image_name)
        if code_check is None or code_check["missing_codes"] or code_check["extra_codes"]:
            continue
        links = [
            link for link in image_link_checks
            if link["status"] == "일치"
            and link["image"] is not None
            and link["image"]["image_name"] == image_name
        ]
        if len(jsons_by_image.get(image_name, set())) != target_count or len(links) != target_count:
            continue
        if not all(bbox_fits_recorded_image(link) for link in links):
            continue

        links = sorted(links, key=lambda item: item["json_path"])
        show_table(
            ["연결 JSON", "image_id", "drug_N", "dl_name", "annotation id", "category_id", "categories.name", "bbox (x, y, width, height)"],
            [
                [
                    link["json_path"],
                    link["image"]["metadata"].get("id"),
                    link["image"]["metadata"].get("drug_N"),
                    link["image"]["metadata"].get("dl_name"),
                    link["annotation"].get("id"),
                    link["annotation"].get("category_id"),
                    category_name_by_json_id.get((link["json_path"], link["annotation"].get("category_id"))),
                    link["annotation"].get("bbox"),
                ]
                for link in links
            ],
        )

        with Image.open(train_image_by_name[image_name]) as image:
            image_rgb = image.convert("RGB")
        fig, ax = plt.subplots(figsize=(9, 11))
        ax.imshow(image_rgb)
        normal_label_color = "#00a651"
        for link in links:
            x, y, box_width, box_height = link["annotation"]["bbox"]
            color = normal_label_color
            ax.add_patch(Rectangle((x, y), box_width, box_height, fill=False, edgecolor=color, linewidth=2.5))
            category_id = link["annotation"].get("category_id")
            category_name = category_name_by_json_id.get((link["json_path"], category_id), "이름 미확인")
            ax.text(
                x, max(0, y - 4), f"{category_id}: {category_name}",
                color="black", fontsize=8, va="bottom",
                bbox={"facecolor": color, "alpha": 0.82, "pad": 2, "edgecolor": "none"},
            )
        ax.set_title(
            f"{image_name}\n"
            f"연결 JSON {len(jsons_by_image[image_name])}개 · 객체 정답 {len(links)}개",
            fontsize=10,
        )
        ax.axis("off")
        plt.tight_layout()
        plt.show()
        return image_name

representative_image_names = {}
for target_count in [3, 4]:
    display(Markdown(f"#### 객체 정답 {target_count}개인 대표 이미지"))
    selected_name = show_representative_image(target_count)
    if selected_name is None:
        display(Markdown(f"조건에 맞는 {target_count}개 객체 사례를 찾지 못했습니다."))
    else:
        representative_image_names[target_count] = selected_name
```

### 4.2 클래스 목록과 원본 ID의 일관성

**목적:** 클래스 수·ID와 이름의 대응·annotation의 클래스 참조를 확인하고 학습용 인덱스의 필요성을 판단한다.

표의 **순번(확인용)**은 약품 목록을 세기 위한 번호이며, 원본 클래스 ID와 구분한다.

```python
names_by_class_id = defaultdict(set)
invalid_category_rows = []
for row in category_rows:
    class_id = row.get("id")
    class_name = row.get("name")
    if (
        not isinstance(class_id, int) or isinstance(class_id, bool)
        or not isinstance(class_name, str) or not class_name.strip()
    ):
        invalid_category_rows.append(row)
        continue
    # 원본 ID와 이름을 그대로 보존한다.
    names_by_class_id[class_id].add(class_name)

annotation_class_ids = {
    row["category_id"] for row in annotation_rows
    if isinstance(row.get("category_id"), int)
    and not isinstance(row.get("category_id"), bool)
}
invalid_annotation_category_count = sum(
    not isinstance(row.get("category_id"), int)
    or isinstance(row.get("category_id"), bool)
    for row in annotation_rows
)
original_class_ids = sorted(names_by_class_id)
class_name_conflicts = {
    class_id: names
    for class_id, names in names_by_class_id.items() if len(names) > 1
}
undefined_class_ids = annotation_class_ids - set(original_class_ids)
is_zero_based_contiguous = original_class_ids == list(range(len(original_class_ids)))
class_catalog = [
    [class_id, " / ".join(sorted(names_by_class_id[class_id]))]
    for class_id in original_class_ids
]

show_table(
    ["확인 항목", "결과"],
    [
        ["categories의 고유 클래스 ID 수", len(original_class_ids)],
        ["annotations에 등장하는 고유 클래스 ID 수", len(annotation_class_ids)],
        ["클래스 ID 범위", f"{min(original_class_ids)} ~ {max(original_class_ids)}" if original_class_ids else "없음"],
        ["0부터 시작하는 연속 ID인가?", is_zero_based_contiguous],
        ["같은 ID에 이름이 여러 개인 경우", len(class_name_conflicts)],
        ["categories에 정의되지 않은 annotation 클래스 ID", len(undefined_class_ids)],
        ["유효한 ID·이름을 확인하지 못한 categories 항목", len(invalid_category_rows)],
        ["유효한 category_id가 없는 annotations 항목", invalid_annotation_category_count],
    ],
)
show_table(
    ["순번(확인용)", "원본 클래스 ID", "약품명(원본 표기)"],
    [[number, *row] for number, row in enumerate(class_catalog, start=1)],
)
if class_name_conflicts:
    show_table(
        ["확인할 클래스 ID", "연결된 이름"],
        [[key, " / ".join(sorted(value))] for key, value in class_name_conflicts.items()],
    )
```

**결과 설명:** 유효한 클래스 ID는 56개이며 1900~41768 범위의 비연속 번호다. annotation에 등장한 클래스 ID는 모두 categories에 정의되어 있었고, 같은 ID에 여러 이름이 연결된 경우는 없었다.


### 4.3 객체별 클래스 참조와 약품 정보의 대응

**목적:** 각 객체가 같은 JSON의 클래스 정의를 가리키는지 검사하고, 연결된 약품 코드·약품명·클래스 ID의 관계를 확인한다.

- 각 annotation의 `category_id`를 **같은 JSON**의 `categories.id`와 연결한다.
- 이미지와 클래스가 각각 하나로 연결된 객체에서 `images.drug_N`과 클래스 ID의 대응을 모은다.
- `images.dl_name`과 `categories.name`은 원문 그대로 비교한다.
- 약품 코드의 숫자부와 클래스 ID가 같은지는 다음 4.4에서 전체 파일을 대상으로 직접 비교한다.

**집계 단위:** 연결·이름 비교는 annotation 항목 수, 여러 값과 연결되는 경우는 고유 약품 코드·클래스 ID 수로 센다.


```python
categories_by_json_id = defaultdict(list)
invalid_local_category_rows = []
for category in category_rows:
    category_id = category.get("id")
    if valid_connection_id(category_id):
        categories_by_json_id[(category["json_path"], category_id)].append(category)
    else:
        invalid_local_category_rows.append(category)

category_link_checks = []
drug_class_links = []
for image_check in image_link_checks:
    annotation = image_check["annotation"]
    category_id = annotation.get("category_id")
    matches = (
        categories_by_json_id.get((annotation["json_path"], category_id), [])
        if valid_connection_id(category_id) else []
    )
    if not valid_connection_id(category_id):
        status = "category_id 누락·형식 확인"
    elif not matches:
        status = "같은 JSON에서 클래스 ID를 찾지 못함"
    elif len(matches) > 1:
        status = "같은 JSON에서 클래스 ID가 여러 항목과 연결됨"
    else:
        status = "일치"
    category_link_checks.append({
        "json_path": annotation["json_path"],
        "annotation": annotation,
        "category_id": category_id,
        "status": status,
    })
    if image_check["status"] == "일치" and status == "일치":
        image = image_check["image"]["metadata"]
        drug_class_links.append({
            "json_path": annotation["json_path"],
            "drug_N": image.get("drug_N"),
            "category_id": category_id,
            "category_name": matches[0].get("name"),
            "drug_name": image.get("dl_name"),
        })

def present_connection_text(value):
    return isinstance(value, str) and bool(value.strip())


class_ids_by_drug_code = defaultdict(set)
drug_codes_by_class_id = defaultdict(set)
drug_names_by_class_id = defaultdict(set)
missing_drug_code_links = []
missing_drug_name_links = []
comparable_drug_names = []
for row in drug_class_links:
    if present_connection_text(row["drug_N"]):
        class_ids_by_drug_code[row["drug_N"]].add(row["category_id"])
        drug_codes_by_class_id[row["category_id"]].add(row["drug_N"])
    else:
        missing_drug_code_links.append(row)
    if present_connection_text(row["drug_name"]):
        drug_names_by_class_id[row["category_id"]].add(row["drug_name"])
    if present_connection_text(row["drug_name"]) and present_connection_text(row["category_name"]):
        comparable_drug_names.append(row)
    else:
        missing_drug_name_links.append(row)

drug_code_class_conflicts = {
    code: ids for code, ids in class_ids_by_drug_code.items() if len(ids) > 1
}
class_drug_code_conflicts = {
    category_id: codes for category_id, codes in drug_codes_by_class_id.items() if len(codes) > 1
}
drug_name_differences = [
    row for row in comparable_drug_names if row["drug_name"] != row["category_name"]
]
category_link_status_counts = Counter(row["status"] for row in category_link_checks)
show_table(
    ["검사 항목", "결과"],
    [
        ["클래스 연결을 검사한 객체 정답 수", len(category_link_checks)],
        ["같은 JSON의 클래스 정의와 연결된 객체 정답 수", category_link_status_counts["일치"]],
        ["id가 없거나 정수가 아닌 categories 항목 수", len(invalid_local_category_rows)],
        ["category_id가 없거나 정수가 아닌 객체 정답 수", category_link_status_counts["category_id 누락·형식 확인"]],
        ["같은 JSON에서 클래스 ID를 찾지 못한 객체 정답 수", category_link_status_counts["같은 JSON에서 클래스 ID를 찾지 못함"]],
        ["같은 JSON에서 클래스 ID가 여러 항목과 연결된 객체 정답 수", category_link_status_counts["같은 JSON에서 클래스 ID가 여러 항목과 연결됨"]],
        ["이미지·클래스가 모두 하나로 연결된 객체 정답 수", len(drug_class_links)],
        ["약품 코드가 없거나 형식 확인이 필요한 연결 항목 수", len(missing_drug_code_links)],
        ["연결된 고유 약품 코드 수", len(class_ids_by_drug_code)],
        ["연결된 고유 클래스 ID 수", len(drug_codes_by_class_id)],
        ["여러 클래스 ID와 연결된 약품 코드 수", len(drug_code_class_conflicts)],
        ["여러 약품 코드와 연결된 클래스 ID 수", len(class_drug_code_conflicts)],
        ["약품명 원문 비교 대상 항목 수", len(comparable_drug_names)],
        ["약품명 결측·형식 문제로 비교하지 못한 항목 수", len(missing_drug_name_links)],
        ["약품명 원문 표기가 다른 항목 수", len(drug_name_differences)],
    ],
)

# 표기 차이·일대다 대응은 검토 대상으로 표시하고 원본 값은 보존한다.
for status, count in sorted(category_link_status_counts.items()):
    if status == "일치":
        continue
    display(Markdown(f"**{status}: 전체 {count}건, 최대 5건 표시**"))
    show_table(
        ["JSON 경로(data/raw 기준)", "annotation ID", "category_id"],
        [
            [row["json_path"], row["annotation"].get("id"), row["category_id"]]
            for row in category_link_checks if row["status"] == status
        ][:5],
    )
if invalid_local_category_rows:
    show_table(
        ["id 형식 확인이 필요한 JSON(최대 5건)", "categories.id"],
        [[row["json_path"], row.get("id")] for row in invalid_local_category_rows[:5]],
    )
if drug_code_class_conflicts:
    show_table(
        ["여러 클래스와 연결된 약품 코드(최대 5개)", "클래스 ID 목록"],
        [[key, sorted(values)] for key, values in sorted(drug_code_class_conflicts.items())[:5]],
    )
if class_drug_code_conflicts:
    show_table(
        ["여러 약품 코드와 연결된 클래스 ID(최대 5개)", "약품 코드 목록"],
        [[key, " / ".join(sorted(values))] for key, values in sorted(class_drug_code_conflicts.items())[:5]],
    )
for label, rows in [
    ("약품 코드 확인 대상", missing_drug_code_links),
    ("약품명 결측·형식 확인 대상", missing_drug_name_links),
    ("약품명 원문 표기 차이", drug_name_differences),
]:
    if rows:
        display(Markdown(f"**{label}: 전체 {len(rows)}건, 최대 5건 표시**"))
        show_table(
            ["JSON 경로(data/raw 기준)", "drug_N", "클래스 ID", "categories.name", "images.dl_name"],
            [[row["json_path"], row["drug_N"], row["category_id"], row["category_name"], row["drug_name"]] for row in rows[:5]],
        )
```

**결과 설명:** annotation 763개 모두 같은 JSON의 클래스 정의와 연결됐다. 약품 코드와 클래스 ID는 각각 56개로 일대일 대응했고, 연결된 약품명과 클래스명 사이의 원문 표기 차이는 없었다.


### 4.4 전체 파일의 약품 코드·클래스·약품명 대응 확인

**목적:** 학습 라벨 JSON 전체에서 약품 코드(`images.drug_N`)와 원본 클래스 ID(`annotations.category_id`)의 대응이 일관적인지 확인하고, 두 위치의 약품명도 비교한다.

**표의 한 행:** 원본 클래스 ID 하나를 나타낸다. 순번은 표를 읽기 위한 번호이며 학습용 ID가 아니다. 값이 여러 종류인 경우 모두 표시한다.

**일치 여부 기준:** 아래 두 조건을 모두 만족하면 `O`, 하나라도 만족하지 않으면 `X`로 표시한다.

1. 약품 코드의 `K-` 뒤 숫자 전체를 정수로 바꾸었을 때 원본 클래스 ID와 같다.
2. 하나의 약품 코드가 하나의 원본 클래스 ID에만 연결되고, 해당 클래스 ID에도 약품 코드가 하나만 연결된다.
3. `categories.name`과 `images.dl_name`의 고유 값이 서로 같다.

판정은 앞선 연결 검사를 통해 연결된 전체 객체 정답을 대상으로 한다. 연결이 누락되거나 값이 비어 있어 비교할 수 없는 경우도 일치로 보지 않는다.


```python
drug_class_catalog = []
for number, category_id in enumerate(original_class_ids, start=1):
    drug_codes = sorted(drug_codes_by_class_id.get(category_id, set()))
    category_names = sorted(names_by_class_id.get(category_id, set()))
    image_drug_names = sorted(drug_names_by_class_id.get(category_id, set()))

    # 약품 코드와 클래스 ID가 서로 일대일로 대응하는지 확인한다.
    code_class_match = (
        len(drug_codes) == 1
        and class_ids_by_drug_code.get(drug_codes[0], set()) == {category_id}
    )
    code_numeric_match = (
        len(drug_codes) == 1
        and drug_codes[0].startswith("K-")
        and drug_codes[0][2:].isdigit()
        and int(drug_codes[0][2:]) == category_id
    )
    # 두 위치의 약품명이 전체 관찰값 기준으로 동일한지 확인한다.
    drug_name_match = (
        bool(category_names)
        and set(category_names) == set(image_drug_names)
    )
    match_status = "O" if code_class_match and code_numeric_match and drug_name_match else "X"

    drug_class_catalog.append([
        number,
        " / ".join(drug_codes) or "연결된 코드 없음",
        category_id,
        " / ".join(category_names) or "categories.name 없음",
        " / ".join(image_drug_names) or "연결된 약품명 없음",
        match_status,
    ])

show_table(
    ["순번", "약품코드", "원본 클래스 ID", "categories.name", "images.dl_name", "일치여부(O/X)"],
    drug_class_catalog,
)
```

### 4.5 클래스별 학습 이미지·객체 수와 불균형

**목적:** 클래스마다 등장하는 고유 학습 이미지 수와 객체 정답 수를 세어 불균형의 크기를 파악한다. 같은 이미지가 여러 JSON에 반복 기록된 경우 이미지 수는 파일명 기준으로 한 번만 센다.

아래는 객체 정답 수가 많은 순서로 정렬한 단일 표 이미지다. 왼쪽 열은 `K-코드 (객체 정답 수)`, 오른쪽 열은 해당 클래스의 모든 유효 bbox crop이다. 괄호 안 숫자는 고유 이미지 수가 아니라 객체 정답 수다.


```python
import statistics
class_objects_for_balance = Counter()
class_images_for_balance = defaultdict(set)
category_status_by_annotation_object = {
    id(row["annotation"]): row["status"] for row in category_link_checks
}

for image_check in image_link_checks:
    annotation = image_check["annotation"]
    category_status = category_status_by_annotation_object.get(id(annotation), "클래스 연결 확인 필요")
    if image_check["status"] != "일치" or category_status != "일치":
        continue
    class_id = annotation.get("category_id")
    if class_id not in names_by_class_id:
        continue
    class_objects_for_balance[class_id] += 1
    image_name = image_check["image"]["image_name"]
    if image_name:
        class_images_for_balance[class_id].add(image_name)

class_balance_rows = [
    {
        "class_id": class_id,
        "class_name": " / ".join(sorted(names_by_class_id[class_id])),
        "image_count": len(class_images_for_balance[class_id]),
        "object_count": class_objects_for_balance[class_id],
    }
    for class_id in original_class_ids
]
class_object_counts = [row["object_count"] for row in class_balance_rows]
class_object_frequency = Counter(class_object_counts)
min_class_objects = min(class_object_counts)
max_class_objects = max(class_object_counts)

show_table(
    ["요약 항목", "결과"],
    [
        ["학습 클래스 수", len(class_balance_rows)],
        ["집계한 annotation 수", sum(class_object_counts)],
        ["클래스별 객체 수 최솟값", min_class_objects],
        ["클래스별 객체 수 중앙값", f"{statistics.median(class_object_counts):g}"],
        ["클래스별 객체 수 최댓값", max_class_objects],
        ["최다/최소 객체 수 비율", f"{max_class_objects / min_class_objects:.1f}배"],
    ],
)

show_table(
    ["클래스당 학습 객체 수", "해당 클래스 수"],
    [[count, n_classes] for count, n_classes in sorted(class_object_frequency.items())],
)

show_table(
    ["원본 클래스 ID", "약품명(원본 표기)", "고유 학습 이미지 수", "객체 정답 수"],
    [
        [row["class_id"], row["class_name"], row["image_count"], row["object_count"]]
        for row in sorted(class_balance_rows, key=lambda row: (-row["object_count"], row["class_id"]))
    ],
)
```

```python
# 클래스별 전체 bbox crop을 두 열 구조의 단일 표 이미지로 그린다.
from math import ceil
from PIL import Image, ImageOps

class_crops = defaultdict(list)
class_uncropped_counts = Counter()
category_status_by_annotation_object = {
    id(row["annotation"]): row["status"] for row in category_link_checks
}

for link in image_link_checks:
    annotation = link["annotation"]
    class_id = annotation.get("category_id")
    image_name = link["image"]["image_name"] if link.get("image") else None
    if (
        link["status"] != "일치"
        or category_status_by_annotation_object.get(id(annotation)) != "일치"
        or class_id not in names_by_class_id
        or image_name not in train_image_by_name
    ):
        continue
    if not bbox_fits_recorded_image(link):
        class_uncropped_counts[class_id] += 1
        continue

    x, y, box_width, box_height = annotation["bbox"]
    metadata = link["image"]["metadata"]
    pad_x, pad_y = box_width * 0.08, box_height * 0.08
    crop_box = (
        max(0, round(x - pad_x)),
        max(0, round(y - pad_y)),
        min(round(x + box_width + pad_x), round(metadata["width"])),
        min(round(y + box_height + pad_y), round(metadata["height"])),
    )
    class_crops[class_id].append({
        "image_name": image_name,
        "json_path": link["json_path"],
        "crop_box": crop_box,
    })

class_rows_desc = sorted(
    class_balance_rows,
    key=lambda row: (-row["object_count"], row["class_id"]),
)
columns = 12
rendered_classes = []
for class_row in class_rows_desc:
    class_id = class_row["class_id"]
    drug_codes = sorted(drug_codes_by_class_id.get(class_id, set()))
    drug_code = " / ".join(drug_codes) if drug_codes else f"클래스 ID {class_id} (약품 코드 미연결)"
    crop_infos = sorted(
        class_crops.get(class_id, []),
        key=lambda crop: (crop["image_name"], crop["json_path"]),
    )
    thumbnails = []
    for crop_info in crop_infos:
        with Image.open(train_image_by_name[crop_info["image_name"]]) as image:
            crop = image.convert("RGB").crop(crop_info["crop_box"])
        thumbnails.append(ImageOps.contain(crop, (190, 110)))
    crop_rows = max(1, ceil(len(thumbnails) / columns))
    rendered_classes.append((class_row, drug_code, thumbnails, crop_rows))

total_body_rows = sum(item[3] for item in rendered_classes)
fig = plt.figure(figsize=(22, max(6, 0.95 * (total_body_rows + 1))), dpi=100)
grid = fig.add_gridspec(
    nrows=total_body_rows + 1,
    ncols=columns + 1,
    width_ratios=[1.7] + [1] * columns,
    hspace=0.025,
    wspace=0.025,
)

def style_table_cell(ax, facecolor="#ffffff"):
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_facecolor(facecolor)
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color("#c9cdd2")
        spine.set_linewidth(0.45)

header_left = fig.add_subplot(grid[0, 0])
style_table_cell(header_left, "#e9edf2")
header_left.text(0.5, 0.5, "K-코드 (약 개수)", ha="center", va="center", fontsize=10, weight="bold", wrap=True)
header_images = fig.add_subplot(grid[0, 1:])
style_table_cell(header_images, "#e9edf2")
header_images.text(0.5, 0.5, "약 이미지 (bbox crop 전체)", ha="center", va="center", fontsize=10, weight="bold")

row_start = 1
for class_row, drug_code, thumbnails, crop_rows in rendered_classes:
    class_id = class_row["class_id"]
    left = fig.add_subplot(grid[row_start:row_start + crop_rows, 0])
    style_table_cell(left, "#f7f8fa")
    left.text(
        0.5, 0.5,
        f"{drug_code} ({class_row['object_count']}개)",
        ha="center", va="center", fontsize=8, weight="bold", wrap=True,
    )
    uncropped = class_uncropped_counts[class_id]
    if uncropped:
        left.text(0.5, 0.12, f"crop 제외 {uncropped}개", ha="center", va="center", fontsize=6, color="#a33")

    for local_row in range(crop_rows):
        for col in range(columns):
            index = local_row * columns + col
            ax = fig.add_subplot(grid[row_start + local_row, col + 1])
            style_table_cell(ax)
            if index < len(thumbnails):
                ax.imshow(thumbnails[index])
            elif not thumbnails and local_row == 0 and col == 0:
                ax.text(0.5, 0.5, "bbox crop 없음", ha="center", va="center", fontsize=7, color="#666")
    row_start += crop_rows

fig.suptitle("약품·클래스별 학습 이미지", fontsize=12, y=0.998)
fig.subplots_adjust(left=0.015, right=0.99, top=0.99, bottom=0.01)
plt.show()
```

**결과 설명:** 56개 클래스의 객체 수는 클래스별로 크게 다르다. 클래스별 객체 수의 최솟값은 3개, 중앙값은 9개, 최댓값은 153개이며, 최다 클래스와 최소 클래스의 차이는 51배다. 아래 표와 클래스 이미지 표는 객체 정답 수가 많은 순으로 정렬했다. 이미지 수는 파일명 기준 고유 이미지 수이며, 객체 수는 annotation 항목 수다.



### 4.6 약코드별 bbox·약품 이미지 불일치 시각 확인

**확인 기준:** 분석 단위는 bbox가 아니라 `K-코드(클래스)`다. 각 행은 한 약코드의 관점에서 `해당 코드로 연결된 bbox crop`과 `같은 코드의 정상 참조 crop`을 비교한다. 빨간색은 4.5에서 확인한 불일치 annotation bbox를 모두 표시하고, 초록색은 같은 이미지에 있는 나머지 bbox다. 같은 원본 이미지에서 두 클래스 bbox가 서로 바뀐 사례는 두 bbox 모두 빨간색으로 표시한다. 오른쪽의 초록색 crop은 해당 K-code의 정상 참조다.

**검토 범위:** 4.5 클래스별 이미지 표를 시각적으로 살펴 발견한 6개 annotation 사례를 정리한다. 이는 자동 검출 결과가 아니라 사람이 이미지와 같은 클래스의 다른 crop을 비교해 선정한 검토 사례다. 이 코드는 원본 라벨을 수정하지 않는다.

### 5.2 필드별 값 분포와 결측 확인

**목적:** 외형·식별 필드별로 기록이 얼마나 채워져 있는지, 실제 값이 얼마나 다양한지 확인해 학습 데이터 검수와 외부 데이터 검색에 쓸 수 있는 정보를 파악한다.

결측은 key 누락·`null`·빈 문자열을 합산한다. 숫자형 치수의 0은 결측과 분리해 별도로 센다. 집계 단위는 5.1과 동일한 약품별 JSON 기록이다.

```python
from collections import Counter
from numbers import Number
import statistics

appearance_distribution_fields = ["chart", "dl_custom_shape", "drug_shape", "form_code_name", "thick", "leng_long", "leng_short", "print_front", "print_back", "color_class1", "color_class2", "line_front", "line_back", "mark_code_front_anal", "mark_code_back_anal", "mark_code_front_img", "mark_code_back_img", "mark_code_front", "mark_code_back"]
appearance_records = field_records["images"]
appearance_distribution_rows = []
numeric_appearance_fields = {"thick", "leng_long", "leng_short"}
for field in appearance_distribution_fields:
    raw_values = [record.get(field) for record in appearance_records]
    present_values = [v for v in raw_values if v is not None and not (isinstance(v, str) and v.strip() == "")]
    missing_count = len(raw_values) - len(present_values)
    unique_count = len({json.dumps(v, ensure_ascii=False, sort_keys=True) for v in present_values})
    if field in numeric_appearance_fields:
        nums = [float(v) for v in present_values if isinstance(v, Number) and not isinstance(v, bool)]
        summary = f"최솟값 {min(nums):g}; 중앙값 {statistics.median(nums):g}; 최댓값 {max(nums):g}; 0 값 {sum(v == 0 for v in nums)}건" if nums else "숫자 값 없음"
    else:
        counts = Counter(str(v).replace("\n", " ") for v in present_values)
        summary = "; ".join(f"{v[:36]} ({n}건)" for v,n in counts.most_common(3)) if counts else "유효 값 없음"
    appearance_distribution_rows.append([field, len(present_values), missing_count, unique_count, summary])
show_table(["필드", "값이 있는 기록", "결측·빈 값", "고유 값 수", "대표 분포 또는 수치 요약"], appearance_distribution_rows)
unique_appearance_filenames = {r.get("file_name") for r in appearance_records if isinstance(r.get("file_name"), str) and r.get("file_name")}
display(Markdown(f"**집계 기준:** 약품별 JSON `images` 기록 {len(appearance_records)}건, 그 안의 고유 `file_name` {len(unique_appearance_filenames)}개. 같은 사진에 여러 약품이 있으면 각 약품 JSON의 메타데이터를 각각 집계한다."))
```

표본은 `images.drug_N`과 같은 JSON의 `annotations.category_id`가 일치하고, 색상·모양·앞면 식별문자 값이 있는 기록에서 색상별로 하나씩 선정한다. 파란 박스는 JSON bbox 위치를 표시하며 정상·오류 판정은 뜻하지 않는다.

### 6.1 촬영 조건별 값 분포와 결측

**목적:** 촬영정보 필드별 제공 값과 결측을 고유 이미지 기준으로 세어, 실제 데이터에서 촬영 조건이 얼마나 다양한지 확인한다. 같은 사진의 여러 JSON 기록은 한 장으로 집계하고, 서로 다른 값이 기록된 경우는 불일치로 표시한다.

```python
from collections import Counter, defaultdict

capture_fields = ["drug_S", "back_color", "light_color", "drug_dir", "camera_la", "camera_lo"]
capture_records_by_filename = defaultdict(list)
for record in field_records["images"]:
    filename = record.get("file_name")
    if filename:
        capture_records_by_filename[filename].append(record)
capture_unique_images = {name: rows[0] for name, rows in capture_records_by_filename.items()}
capture_distribution_rows = []
for field in capture_fields:
    values = [record.get(field) for record in capture_unique_images.values()]
    present = [value for value in values if value is not None and not (isinstance(value, str) and value.strip() == "")]
    counts = Counter(str(value) for value in present)
    conflicts = sum(len({str(row.get(field)) for row in rows}) > 1 for rows in capture_records_by_filename.values())
    distribution = "; ".join(f"{value} ({count}장)" for value, count in sorted(counts.items(), key=lambda x: (-x[1], x[0])))
    capture_distribution_rows.append([field, len(present), len(values)-len(present), conflicts, distribution or "유효 값 없음"])
show_table(["필드", "값이 있는 이미지", "결측 이미지", "같은 이미지 내 불일치", "관찰값 분포"], capture_distribution_rows)
display(Markdown(f"**집계 단위:** 고유 학습 이미지 {len(capture_unique_images)}장. 중복 `images` 기록 {sum(len(rows)-1 for rows in capture_records_by_filename.values())}건은 사진 파일명으로 묶었다."))
```

**결과 설명:** 촬영 필드에는 정상알약 상태, 연회색 배경, 주백색 조명, 앞면·뒷면 방향이 기록됐다. `camera_la`는 70·75·90, `camera_lo`는 0으로 관찰됐다.


### 6.2 촬영 메타데이터와 실제 이미지 대조

**목적:** 같은 이미지에 반복 기록된 촬영정보가 일관적인지 확인하고, `camera_la` 두 값 조합별 2장 묶음과 70·75·90 세 값이 모두 있는 3장 묶음 사례를 실제 이미지로 비교한다.

파란 표시는 없으며, 이 절의 이미지는 촬영 묶음 후보를 육안으로 확인하기 위한 예시다. `camera_la`·`camera_lo`는 Kaggle 표의 명칭만 확인되었으므로 단위나 물리적 의미를 추정하지 않는다.

```python
show_metadata_consistency({"back_color", "light_color", "drug_dir", "camera_la", "camera_lo"})
```

```python
from PIL import Image as PILImage, ImageDraw, ImageFont
from collections import defaultdict
from pathlib import Path

# 파일명 끝의 camera_la, camera_lo, size가 JSON 메타데이터와 맞는 사진만 묶는다.
# 이 셀에서 사용하는 이미지 파일명→경로 사전을 직접 만든다.
train_image_by_name_for_capture = {path.name: path for path in train_image_paths}
capture_groups_for_visual = defaultdict(list)
for filename, record in capture_unique_images.items():
    parts = filename.rsplit(".", 1)[0].split("_")
    if len(parts) < 4:
        continue
    suffix = parts[-3:]
    try:
        matches = (int(suffix[0]) == int(record.get("camera_la")) and int(suffix[1]) == int(record.get("camera_lo")) and int(suffix[2]) == int(record.get("size")))
    except (TypeError, ValueError):
        matches = False
    if matches:
        base = "_".join(parts[:-3])
        capture_groups_for_visual[base].append((filename, record))

# camera_la 두 값 조합별 2장 묶음 하나씩, 70·75·90 전체가 있는 3장 묶음 3개를 고른다.
pair_combinations = [(70, 75), (70, 90), (75, 90)]
selected_pair_groups = []
for combination in pair_combinations:
    match_group = None
    for base, group in sorted(capture_groups_for_visual.items()):
        angles = tuple(sorted(int(record["camera_la"]) for _, record in group))
        if len(group) == 2 and angles == combination:
            match_group = (base, sorted(group, key=lambda item: int(item[1]["camera_la"])))
            break
    if match_group is None:
        raise ValueError(f"camera_la 조합을 가진 2장 묶음을 찾지 못했습니다: {combination}")
    selected_pair_groups.append(match_group)

selected_triple_groups = []
for base, group in sorted(capture_groups_for_visual.items()):
    angles = {int(record["camera_la"]) for _, record in group}
    if len(group) == 3 and angles == {70, 75, 90}:
        selected_triple_groups.append((base, sorted(group, key=lambda item: int(item[1]["camera_la"]))))
selected_triple_groups = selected_triple_groups[:3]
if len(selected_triple_groups) < 3:
    raise ValueError("70·75·90 세 값이 있는 3장 묶음을 3개 찾지 못했습니다.")

selected_capture_groups = [(f"Pair {a}-{b}", base, group) for (a,b), (base,group) in zip(pair_combinations, selected_pair_groups)]
selected_capture_groups += [(f"Triplet {i}", base, group) for i,(base,group) in enumerate(selected_triple_groups,1)]
capture_visual_table_rows = []
for group_label, base, group in selected_capture_groups:
    for filename, record in group:
        capture_visual_table_rows.append([group_label, base, filename, record.get("camera_la"), record.get("camera_lo"), record.get("drug_dir")])
show_table(["비교 묶음", "공통 파일명 접두부", "이미지 파일명", "camera_la", "camera_lo", "drug_dir"], capture_visual_table_rows)

font = ImageFont.load_default()
def make_capture_sheet(groups, columns, card_width=340, card_height=350):
    gap, header_height = 12, 44
    row_count = len(groups)
    sheet = PILImage.new("RGB", (columns*card_width+(columns+1)*gap, row_count*(card_height+header_height)+(row_count+1)*gap), "white")
    draw = ImageDraw.Draw(sheet)
    for row_index, (group_label, base, group) in enumerate(groups):
        y0 = gap + row_index*(card_height+header_height+gap)
        draw.text((gap, y0+5), f"{group_label} | {base}", font=font, fill=(25,25,25))
        for col_index, (filename, record) in enumerate(group):
            x0 = gap + col_index*(card_width+gap)
            image_y = y0+header_height
            draw.rectangle((x0,image_y,x0+card_width-1,image_y+card_height-1),outline=(205,210,215),width=2)
            draw.text((x0+7,image_y+6),f"camera_la={record.get('camera_la')}",font=font,fill=(25,25,25))
            with PILImage.open(train_image_by_name_for_capture[filename]) as source:
                image=source.convert("RGB")
            image.thumbnail((card_width-18,card_height-38),PILImage.Resampling.LANCZOS)
            px=x0+(card_width-image.width)//2
            py=image_y+28+(card_height-34-image.height)//2
            sheet.paste(image,(px,py))
    return sheet

pair_sheet = make_capture_sheet(selected_capture_groups[:3], columns=2)
triple_sheet = make_capture_sheet(selected_capture_groups[3:], columns=3)
display(Markdown("**camera_la 2장 조합별 사례**"))
display(pair_sheet)
display(Markdown("**70·75·90 세 값이 모두 있는 3장 묶음 사례 3개**"))
display(triple_sheet)
```

### 6.3 촬영 묶음과 학습·검증 분할 검토

**목적:** 파일명 접미부의 `camera_la`·`camera_lo`·`size` 값이 JSON 메타데이터와 일치하는지 확인하고, 접미부를 제외한 같은 파일명 접두부의 여러 촬영 이미지를 같은 분할 단위로 묶을 근거가 있는지 집계한다. 이는 파일명·필드의 관찰 관계를 이용한 분할 후보 분석이며, 숫자의 물리적 의미를 해석하지 않는다.

```python
from collections import Counter, defaultdict
from pathlib import Path

capture_view_groups = defaultdict(list)
capture_suffix_validation = []
for filename, record in capture_unique_images.items():
    parts = Path(filename).stem.split("_")
    if len(parts) < 4:
        capture_suffix_validation.append([filename, "접미부 구조 확인 불가"])
        continue
    suffix = parts[-3:]
    try:
        matches = int(suffix[0]) == int(record.get("camera_la")) and int(suffix[1]) == int(record.get("camera_lo")) and int(suffix[2]) == int(record.get("size"))
    except (TypeError, ValueError):
        matches = False
    if matches:
        capture_view_groups["_".join(parts[:-3])].append((filename, record))
    else:
        capture_suffix_validation.append([filename, "파일명 접미부와 JSON 값 불일치"] )
group_size_counts = Counter(len(group) for group in capture_view_groups.values())
images_in_multi_view_groups = sum(len(group) for group in capture_view_groups.values() if len(group) > 1)
angle_sets = Counter(tuple(sorted(int(record["camera_la"]) for _, record in group)) for group in capture_view_groups.values() if len(group) > 1)
show_table(["확인 항목", "결과"], [["고유 이미지", len(capture_unique_images)], ["파일명 접미부와 JSON 값 일치", f"{len(capture_unique_images)-len(capture_suffix_validation)}/{len(capture_unique_images)}"], ["접두부 묶음 수", len(capture_view_groups)], ["3장 묶음 수 (70·75·90)", sum(len(g)==3 and {int(r["camera_la"]) for _,r in g}=={70,75,90} for g in capture_view_groups.values())], ["2장 묶음 수", group_size_counts.get(2,0)], ["1장만 있는 묶음 수", group_size_counts.get(1,0)], ["2장 이상 묶음에 포함된 이미지", images_in_multi_view_groups]])
if capture_suffix_validation:
    show_table(["파일명", "확인 사항"], capture_suffix_validation[:10])
show_table(["2장 이상 묶음의 camera_la 값 조합", "묶음 수"], [[", ".join(map(str, angles)), count] for angles,count in sorted(angle_sets.items())])
```

**필요한 조치:** 분할 시 접미부를 제외한 파일명 접두부 묶음을 같은 train 또는 validation에 둔다. 배경·조명 보정은 현재 관찰만으로 필요성을 단정하지 않고, 증강 실험에서 약한 밝기·대비 변화를 별도 비교한다. 뒷면 방향은 그룹 단위 분할을 유지하면서 검증셋에 포함될 수 있는지 확인하고, 적은 표본 수라는 한계를 함께 보고한다. 카메라 필드 숫자의 단위·좌표 의미는 미확정으로 둔다.

### 7.1 bbox 형식과 기본 유효성

**목적:** Kaggle에서 안내한 `[x, y, width, height]` 형식에 맞춰 bbox의 원소 수·숫자 여부·너비와 높이의 양수 여부를 전체 annotation에서 확인한다.

```python
import math
# 전체 학습 annotation의 bbox를 모으고 이미지 정보와 연결한다.
box_records = []
for json_path in json_paths:
    with json_path.open(encoding="utf-8-sig") as f:
        document = json.load(f)
    image_by_id = {image.get("id"): image for image in document.get("images", [])}
    category_by_id = {category.get("id"): category.get("name") for category in document.get("categories", [])}
    for annotation in document.get("annotations", []):
        image = image_by_id.get(annotation.get("image_id"))
        if image is None and len(document.get("images", [])) == 1:
            image = document["images"][0]
        if image is not None:
            box_records.append({"file_name": image.get("file_name"), "image_width": image.get("width"), "image_height": image.get("height"), "annotation_id": annotation.get("id"), "category_id": annotation.get("category_id"), "category_name": category_by_id.get(annotation.get("category_id")), "bbox": annotation.get("bbox"), "area": annotation.get("area")})

def bbox_issues(record):
    bbox = record.get("bbox")
    if not isinstance(bbox, list) or len(bbox) != 4:
        return ["bbox 형식"]
    if any(not isinstance(value, (int, float)) or isinstance(value, bool) or not math.isfinite(value) for value in bbox):
        return ["bbox 숫자값"]
    x, y, width, height = bbox
    result = []
    if width <= 0 or height <= 0: result.append("너비/높이 0 이하")
    iw, ih = record.get("image_width"), record.get("image_height")
    if not isinstance(iw, (int, float)) or not isinstance(ih, (int, float)): result.append("이미지 크기 없음")
    elif x < 0 or y < 0 or x + width > iw or y + height > ih: result.append("이미지 경계 밖")
    area = record.get("area")
    if isinstance(area, (int, float)) and width > 0 and height > 0 and area != width * height: result.append("area 불일치")
    return result

format_ok = sum(isinstance(r["bbox"], list) and len(r["bbox"]) == 4 and all(isinstance(v, (int, float)) and not isinstance(v, bool) and math.isfinite(v) for v in r["bbox"]) for r in box_records)
positive = sum(isinstance(r["bbox"], list) and len(r["bbox"]) == 4 and r["bbox"][2] > 0 and r["bbox"][3] > 0 for r in box_records)
show_table(["확인 항목", "결과"], [["annotation 수", len(box_records)], ["bbox가 숫자 4개", f"{format_ok}/{len(box_records)}"], ["width·height 양수", f"{positive}/{len(box_records)}"], ["bbox 해석", "[x, y, width, height] (Kaggle 안내)"]])
```

**결과 설명:** 전체 763개 annotation의 `bbox`는 모두 숫자 4개이고 너비·높이가 양수다. 따라서 현재 형식은 학습용 `[x, y, width, height]` 변환을 진행할 수 있는 형태다. 형식 통과가 박스 위치나 약품 클래스의 정답성을 보장하지는 않는다.

### 7.2 이미지 경계와 area 일관성

**목적:** 박스가 이미지 내부에 있는지 확인하고, JSON의 `area`가 `width × height`와 일치하는지 비교해 학습 전 검토 후보를 찾는다.

```python
boundary_candidates = [r for r in box_records if "이미지 경계 밖" in bbox_issues(r)]
area_candidates = [r for r in box_records if "area 불일치" in bbox_issues(r)]
review_candidates = [r for r in box_records if bbox_issues(r)]
show_table(["검사 항목", "통과", "검토 후보"], [["이미지 경계", f"{len(box_records)-len(boundary_candidates)}/{len(box_records)}", len(boundary_candidates)], ["area = bbox 너비×높이", f"{len(box_records)-len(area_candidates)}/{len(box_records)}", len(area_candidates)], ["검토 후보 이미지 수", "—", len({r["file_name"] for r in review_candidates})]])
rows=[]
for r in review_candidates:
    b=r["bbox"]
    rows.append([r["file_name"],r["annotation_id"],r["category_id"],b,f"{r['image_width']}×{r['image_height']}",r["area"],b[2]*b[3],", ".join(bbox_issues(r))])
show_table(["이미지 파일", "annotation_id", "category_id", "bbox", "이미지 크기", "JSON area", "bbox w×h", "검토 이유"], rows)
```

**결과 설명:** 이미지 경계 밖 박스 1개와 `area` 불일치 6개가 확인됐다. 검토 후보 annotation은 7건이며, 같은 이미지에 두 후보 박스가 있는 경우를 합치면 6개 이미지다. `area` 불일치가 곧 bbox 오답이라는 뜻은 아니므로 시각 확인 전 자동 수정하지 않는다.

### 7.3 bbox 위치·크기 분포

**목적:** 이미지 경계 안에 있는 박스만 대상으로 픽셀 크기와 이미지 대비 점유 비율·중심 위치를 요약해 학습 입력 크기와 작은 객체 대응을 검토한다.

```python
in_bounds_records = [r for r in box_records if not any(i in bbox_issues(r) for i in ["bbox 형식", "bbox 숫자값", "너비/높이 0 이하", "이미지 크기 없음", "이미지 경계 밖"])]
def quantile(values, p):
    values=sorted(values); pos=(len(values)-1)*p; lo=int(pos); hi=min(lo+1,len(values)-1); f=pos-lo
    return values[lo]*(1-f)+values[hi]*f
def summary(values, digits=1): return [round(min(values),digits),round(quantile(values,.5),digits),round(quantile(values,.9),digits),round(max(values),digits)]
widths=[r["bbox"][2] for r in in_bounds_records]; heights=[r["bbox"][3] for r in in_bounds_records]
fractions=[r["bbox"][2]*r["bbox"][3]/(r["image_width"]*r["image_height"])*100 for r in in_bounds_records]
center_x=[(r["bbox"][0]+r["bbox"][2]/2)/r["image_width"] for r in in_bounds_records]
center_y=[(r["bbox"][1]+r["bbox"][3]/2)/r["image_height"] for r in in_bounds_records]
show_table(["분포 항목", "최솟값", "중앙값", "90백분위", "최댓값", "단위"], [["bbox 너비",*summary(widths),"px"],["bbox 높이",*summary(heights),"px"],["이미지 대비 bbox 면적",*summary(fractions,2),"%"],["bbox 중심 x 위치",*summary(center_x,3),"이미지 너비 비율"],["bbox 중심 y 위치",*summary(center_y,3),"이미지 높이 비율"]])
display(Markdown(f"**집계 대상:** 경계 안에 있고 크기가 양수인 bbox {len(in_bounds_records)}건."))
```

**결과 설명:** 위치·크기 분포는 이미지 경계 검사를 통과한 박스만 집계한다. 픽셀 크기와 이미지 대비 점유 비율은 모델 입력 크기나 작은 객체 증강을 논의할 근거가 되지만, 현재 분포만으로 특정 리사이즈·증강 강도를 확정하지 않는다.

### 7.4 서로 다른 클래스 bbox 간 겹침 확인

**목적:** 같은 이미지의 annotation끼리 bbox 교차 영역과 IoU를 계산한다. 겹치는 서로 다른 클래스 쌍은 빨간색으로, 같은 이미지의 나머지 bbox는 초록색으로 표시해 전체 맥락에서 확인한다. 빨간색은 겹침 또는 경계·area 검토 후보, 초록색은 이 두 검사에서 후보가 아니라는 뜻이다. 겹침만으로 라벨 오류를 확정하지 않는다.

```python
from collections import defaultdict
from itertools import combinations
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont

# 한 이미지에 연결된 annotation을 모아, 서로 다른 category_id의 bbox 쌍을 비교한다.
records_by_image = defaultdict(list)
for record in box_records:
    records_by_image[record["file_name"]].append(record)
class_overlap_pairs = []
for file_name, records in records_by_image.items():
    for first, second in combinations(records, 2):
        if first.get("category_id") == second.get("category_id"):
            continue
        x1,y1,w1,h1=first["bbox"]; x2,y2,w2,h2=second["bbox"]
        intersection_width=max(0,min(x1+w1,x2+w2)-max(x1,x2))
        intersection_height=max(0,min(y1+h1,y2+h2)-max(y1,y2))
        intersection_area=intersection_width*intersection_height
        if intersection_area == 0:
            continue
        union_area=w1*h1+w2*h2-intersection_area
        class_overlap_pairs.append({"file_name":file_name,"first":first,"second":second,"intersection_area":intersection_area,"iou":intersection_area/union_area if union_area else 0})
class_overlap_pairs.sort(key=lambda row:(row["file_name"],-row["iou"]))
rows=[]
for pair in class_overlap_pairs:
    a,b=pair["first"],pair["second"]
    rows.append([pair["file_name"],a["category_id"],a["bbox"],b["category_id"],b["bbox"],pair["intersection_area"],round(pair["iou"],4)])
show_table(["이미지 파일","클래스 ID 1","bbox 1","클래스 ID 2","bbox 2","교차 면적(px²)","IoU"],rows)
display(Markdown(f"**결과 요약:** 서로 다른 클래스 bbox가 겹치는 쌍 {len(class_overlap_pairs)}개, 해당 이미지 {len({p['file_name'] for p in class_overlap_pairs})}장. 겹침 자체는 오류 확정이 아니며, IoU가 1인 동일 좌표 쌍은 우선 검토 대상으로 본다."))

# 전체 bbox를 표시해 문맥을 보여준다. 겹치는 클래스 쌍 또는 경계/area 후보는 빨강, 나머지는 초록이다.
overlap_record_ids={id(record) for pair in class_overlap_pairs for record in (pair["first"],pair["second"])}
image_by_name={path.name:path for path in train_image_paths}
font_paths=["/usr/share/fonts/truetype/nanum/NanumGothic.ttf","/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc","/System/Library/Fonts/AppleSDGothicNeo.ttc","/System/Library/Fonts/Supplemental/AppleGothic.ttf"]
font_path=next((path for path in font_paths if Path(path).is_file()),None)
try: font=ImageFont.truetype(font_path,14) if font_path else ImageFont.load_default()
except OSError: font=ImageFont.load_default()
files=sorted({pair["file_name"] for pair in class_overlap_pairs})
card_w,card_h=720,560
overlap_sheet=Image.new("RGB",(card_w*2,card_h*3+48),"white")
draw=ImageDraw.Draw(overlap_sheet)
draw.text((14,12),"빨강: 서로 다른 클래스 간 겹침 또는 경계·area 후보 | 초록: 두 검사에서 검토 후보 아님",fill=(30,30,30),font=font)
for index,file_name in enumerate(files):
    x0,y0=(index%2)*card_w,48+(index//2)*card_h
    draw.rectangle((x0+4,y0+4,x0+card_w-5,y0+card_h-5),outline=(190,190,190))
    pairs=[pair for pair in class_overlap_pairs if pair["file_name"]==file_name]
    iou_text=", ".join(f"{pair['iou']:.4f}" for pair in pairs)
    draw.text((x0+12,y0+8),f"{file_name} | 겹침 쌍 IoU: {iou_text}",fill=(20,20,20),font=font)
    image_path=image_by_name.get(file_name)
    if image_path is None: continue
    image=Image.open(image_path).convert("RGB"); original_size=image.size
    image.thumbnail((card_w-26,card_h-38))
    ix=x0+(card_w-image.width)//2; iy=y0+32+(card_h-38-image.height)//2
    overlap_sheet.paste(image,(ix,iy)); painter=ImageDraw.Draw(overlap_sheet)
    sx,sy=image.width/original_size[0],image.height/original_size[1]
    for record in records_by_image[file_name]:
        bbox=record.get("bbox")
        if not isinstance(bbox,list) or len(bbox)!=4 or not all(isinstance(v,(int,float)) for v in bbox): continue
        x,y,w,h=bbox; iw,ih=record["image_width"],record["image_height"]
        flagged=id(record) in overlap_record_ids or bool(bbox_issues(record))
        color=(220,35,35) if flagged else (20,155,65)
        left,top,right,bottom=max(0,x),max(0,y),min(iw,x+w),min(ih,y+h)
        label=f"class {record['category_id']}"+(" *" if id(record) in overlap_record_ids else "")
        if right>left and bottom>top:
            painter.rectangle((ix+left*sx,iy+top*sy,ix+right*sx,iy+bottom*sy),outline=color,width=4)
            painter.text((ix+left*sx,max(iy,iy+top*sy-17)),label,fill=color,font=font,stroke_width=1,stroke_fill="white")
        else:
            edge_x=ix+image.width-8 if x>=iw else ix+3; edge_y=iy+image.height//2
            painter.ellipse((edge_x-6,edge_y-6,edge_x+6,edge_y+6),fill=color)
            painter.text((ix+4,iy+4),label+" outside",fill=color,font=font,stroke_width=1,stroke_fill="white")
display(overlap_sheet)
```

**결과 설명:** 5개 이미지에서 서로 다른 클래스 bbox 겹침 쌍 5개가 확인됐다. 그중 3쌍은 두 bbox 좌표가 완전히 같아 IoU가 1.0이고, 나머지 2쌍은 작은 영역만 겹친다. 동일 좌표의 서로 다른 클래스 라벨은 중복 라벨 또는 클래스 연결 오류 가능성이 높아 우선 검토하되, 실제 이미지와 클래스 의미를 확인한 뒤 수정 여부를 정한다.

### 7.5 경계·area 검토 후보 이미지 시각화

**목적:** 7.2에서 찾은 경계·area 검토 후보 이미지의 전체 박스를 함께 그려 경계·면적 이상을 눈으로 확인한다. 각 이미지 위에는 후보별 `bbox`, JSON `area`, `bbox w×h`, 검토 이유를 표시한다. 초록색은 경계와 `area` 검사를 통과한 박스, 빨간색은 둘 중 하나라도 검토가 필요한 박스이며 클래스 일치 여부를 뜻하지 않는다.

```python
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont
from IPython.display import display
import textwrap

candidate_names = sorted({r["file_name"] for r in box_records if bbox_issues(r)})
image_by_name = {path.name: path for path in train_image_paths}
# Korean-capable fonts for Kaggle Linux and local macOS environments.
font_candidates = ["/usr/share/fonts/truetype/nanum/NanumGothic.ttf", "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc", "/System/Library/Fonts/AppleSDGothicNeo.ttc", "/System/Library/Fonts/Supplemental/AppleGothic.ttf"]
font_path = next((path for path in font_candidates if Path(path).is_file()), None)
try:
    body_font = ImageFont.truetype(font_path, 13) if font_path else ImageFont.load_default()
    title_font = ImageFont.truetype(font_path, 15) if font_path else ImageFont.load_default()
except OSError:
    body_font = title_font = ImageFont.load_default()

card_w, card_h = 720, 620
contact_sheet = Image.new("RGB", (card_w*2, card_h*3+52), "white")
draw = ImageDraw.Draw(contact_sheet)
draw.text((15,12), "빨강: 검토 후보 | 초록: 경계·area 검사 통과 (클래스 일치 의미 아님)", fill=(25,25,25), font=title_font)
for index, file_name in enumerate(candidate_names):
    x0, y0 = (index%2)*card_w, 52+(index//2)*card_h
    draw.rectangle((x0+4,y0+4,x0+card_w-5,y0+card_h-5),outline=(185,185,185),width=1)
    draw.text((x0+12,y0+8),file_name,fill=(20,20,20),font=title_font)
    review_rows=[r for r in box_records if r["file_name"]==file_name and bbox_issues(r)]
    detail_y=y0+32
    for record in review_rows:
        bbox=record.get("bbox"); box_area=bbox[2]*bbox[3] if isinstance(bbox,list) and len(bbox)==4 else "확인 불가"
        detail=f"bbox={bbox} | JSON area={record.get('area')} | bbox w×h={box_area} | 검토 이유={', '.join(bbox_issues(record))}"
        for line in textwrap.wrap(detail,width=82,break_long_words=True,break_on_hyphens=False):
            draw.text((x0+12,detail_y),line,fill=(190,35,35),font=body_font); detail_y+=18
    image_path=image_by_name.get(file_name); image_top=max(detail_y+8,y0+94)
    if image_path is None: continue
    image=Image.open(image_path).convert("RGB"); original_size=image.size
    available_h=max(100,y0+card_h-12-image_top); image.thumbnail((card_w-28,available_h))
    image_x=x0+(card_w-image.width)//2; image_y=image_top+(available_h-image.height)//2
    contact_sheet.paste(image,(image_x,image_y)); painter=ImageDraw.Draw(contact_sheet)
    sx,sy=image.width/original_size[0],image.height/original_size[1]
    for record in [r for r in box_records if r["file_name"]==file_name]:
        bbox=record.get("bbox"); flagged=bool(bbox_issues(record)); color=(220,35,35) if flagged else (20,155,65)
        if not isinstance(bbox,list) or len(bbox)!=4: continue
        x,y,w,h=bbox; iw,ih=record["image_width"],record["image_height"]
        left,top,right,bottom=max(0,x),max(0,y),min(iw,x+w),min(ih,y+h)
        if right>left and bottom>top:
            painter.rectangle((image_x+left*sx,image_y+top*sy,image_x+right*sx,image_y+bottom*sy),outline=color,width=3)
            painter.text((image_x+left*sx,max(image_y,image_y+top*sy-14)),str(record["category_id"])+(" !" if flagged else ""),fill=color,font=body_font,stroke_width=1,stroke_fill="white")
        else:
            edge_x=image_x+image.width-8 if x>=iw else image_x+3; edge_y=image_y+image.height//2
            painter.ellipse((edge_x-6,edge_y-6,edge_x+6,edge_y+6),fill=color)
            painter.text((image_x+4,image_y+4),f"경계 밖 bbox | category_id={record['category_id']}",fill=color,font=body_font,stroke_width=1,stroke_fill="white")
bbox_geometry_sheet = contact_sheet
display(bbox_geometry_sheet)
show_table(["검토 후보 이미지 수", "검토 후보 annotation 수"], [[len(candidate_names),sum(bool(bbox_issues(r)) for r in box_records)]])
```

### 7.6 학습 준비를 위한 결과와 조치

**관찰한 사실:** bbox 763건은 모두 Kaggle에서 안내한 4개 숫자 형식이고 너비·높이가 양수다. 서로 다른 클래스 bbox 겹침 5쌍이 5개 이미지에서 확인됐으며, 이 중 3쌍은 bbox 좌표가 완전히 같아 IoU가 1.0이다. 경계 밖 박스 1건과 `area` 불일치 6건도 확인됐다. 경계 안의 762개 박스에서 너비 중앙값은 230px, 높이 중앙값은 227px, 이미지 대비 박스 면적 중앙값은 약 4.29%다.

**학습에 미치는 영향:** 서로 다른 클래스가 동일한 bbox를 공유하면 한 객체에 상충하는 정답이 학습될 가능성이 있다. 작은 면적의 bbox 겹침은 실제 객체가 가까이 놓여 생길 수도 있어 겹침만으로 오류를 확정하지 않는다. 경계 밖 bbox는 좌표 변환·잘림 문제를 일으킬 수 있고, `area` 불일치는 해당 필드를 사용하는 처리에서 혼선을 줄 수 있다. 클래스 정체성은 4번의 이미지·약품 대조 결과와 함께 판단한다.

**필요한 조치:** 7.4의 겹침 쌍, 특히 IoU 1.0인 서로 다른 클래스 bbox 3쌍은 이미지와 약품 정보를 대조해 라벨 중복인지 클래스 연결 오류인지 확인한다. 7.5의 경계·area 후보도 시각 검토한다. 원본 라벨은 유지하고, 수정이 확정된 경우에만 학습용 변환본에 반영한다. 모델 입력 크기·작은 객체 증강은 분포를 근거로 검증 실험에서 비교한다.

```python
from pathlib import Path
from collections import defaultdict
from PIL import Image as PILImage, ImageDraw, ImageFont
from IPython.display import display
import math

review_image_names = ['K-001900-016548-019607-033009_0_2_0_2_70_000_200.png', 'K-002483-012081-012778-025438_0_2_0_2_90_000_200.png', 'K-003351-013900-021325_0_2_0_2_70_000_200.png', 'K-003351-013900-036637_0_2_0_2_70_000_200.png', 'K-003351-016262-018357_0_2_0_2_75_000_200.png', 'K-003351-019232-029667_0_2_1_2_70_000_200.png', 'K-003351-019232-035206_0_2_0_2_90_000_200.png', 'K-003351-020014-020238_0_2_0_2_90_000_200.png', 'K-003351-020014-022074_0_2_0_2_90_000_200.png', 'K-003351-020238-031863_0_2_0_2_70_000_200.png', 'K-003351-021325-032310_0_2_0_2_90_000_200.png', 'K-003351-021325-033880_0_2_0_2_70_000_200.png', 'K-003351-029667-031863_0_2_0_2_70_000_200.png', 'K-003351-032310-038162_0_2_0_2_70_000_200.png', 'K-003351-033880-035206_0_2_0_2_70_000_200.png', 'K-003351-033880-035206_0_2_0_2_90_000_200.png', 'K-003351-033880-038162_0_2_0_2_70_000_200.png', 'K-003351-033880-038162_0_2_0_2_75_000_200.png', 'K-003351-035206-041768_0_2_0_2_70_000_200.png', 'K-003483-025367-027733-035206_0_2_0_2_75_000_200.png', 'K-003544-004543-012247-016548_0_2_0_2_90_000_200.png']
image_by_name = {path.name: path for path in train_image_paths}
records_by_review_image = defaultdict(list)
for record in box_records:
    if record["file_name"] in review_image_names:
        records_by_review_image[record["file_name"]].append(record)

# 기존 4.6의 클래스·약품 불일치 후보
class_mismatch_keys = {(case["image_name"], case["category_id"]) for case in class_mismatch_cases}
# 7.4의 서로 다른 클래스 bbox 겹침 후보
overlap_annotation_keys = {
    (pair["file_name"], record.get("annotation_id"))
    for pair in class_overlap_pairs
    for record in (pair["first"], pair["second"])
}
# 7.2의 bbox 경계·area 검토 후보
geometry_annotation_keys = {
    (record["file_name"], record.get("annotation_id"))
    for record in box_records if bbox_issues(record)
}

font_candidates = [
    "/usr/share/fonts/truetype/nanum/NanumGothic.ttf",
    "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc",
    "/System/Library/Fonts/AppleSDGothicNeo.ttc",
    "/System/Library/Fonts/Supplemental/AppleGothic.ttf",
]
font_path = next((path for path in font_candidates if Path(path).is_file()), None)
try:
    title_font = ImageFont.truetype(font_path, 17) if font_path else ImageFont.load_default()
    label_font = ImageFont.truetype(font_path, 13) if font_path else ImageFont.load_default()
except OSError:
    title_font = label_font = ImageFont.load_default()

card_w, card_h = 520, 610
page_size = 6
for page_start in range(0, len(review_image_names), page_size):
    page_names = review_image_names[page_start:page_start + page_size]
    columns = 3
    rows = math.ceil(len(page_names) / columns)
    sheet = PILImage.new("RGB", (card_w * columns, card_h * rows + 48), "white")
    painter = ImageDraw.Draw(sheet)
    painter.text((14, 12), f"검토 이미지 {page_start + 1:02d}–{page_start + len(page_names):02d} / {len(review_image_names)}", fill=(25,25,25), font=title_font)
    for offset, file_name in enumerate(page_names):
        index = page_start + offset + 1
        x0 = (offset % columns) * card_w
        y0 = 48 + (offset // columns) * card_h
        painter.rectangle((x0+4,y0+4,x0+card_w-5,y0+card_h-5), outline=(190,190,190), width=1)
        painter.text((x0+10,y0+8), f"{index:02d}. {file_name}", fill=(20,20,20), font=label_font)
        image_path = image_by_name.get(file_name)
        if image_path is None:
            painter.text((x0+12,y0+40), "이미지 파일을 찾지 못했습니다.", fill=(200,35,35), font=label_font)
            continue
        image = PILImage.open(image_path).convert("RGB")
        original_w, original_h = image.size
        image.thumbnail((card_w-22, card_h-52), PILImage.Resampling.LANCZOS)
        image_x = x0 + (card_w-image.width)//2
        image_y = y0 + 34 + (card_h-42-image.height)//2
        sheet.paste(image,(image_x,image_y))
        draw = ImageDraw.Draw(sheet)
        sx, sy = image.width/original_w, image.height/original_h
        for record in records_by_review_image[file_name]:
            bbox = record.get("bbox")
            if not isinstance(bbox,list) or len(bbox)!=4 or not all(isinstance(v,(int,float)) for v in bbox):
                continue
            x,y,w,h=bbox
            red = ((file_name,record.get("category_id")) in class_mismatch_keys
                   or (file_name,record.get("annotation_id")) in overlap_annotation_keys
                   or (file_name,record.get("annotation_id")) in geometry_annotation_keys)
            color=(220,35,35) if red else (20,155,65)
            left=image_x+x*sx; top=image_y+y*sy
            right=image_x+(x+w)*sx; bottom=image_y+(y+h)*sy
            draw.rectangle((left,top,right,bottom),outline=color,width=3)
            label=f"{record.get('category_id')}" + (" !" if red else "")
            draw.text((left,max(image_y,top-15)),label,fill=color,font=label_font,stroke_width=1,stroke_fill="white")
        candidate = missing_object_candidates.get(file_name)
        if candidate:
            x,y,w,h=candidate["bbox"]
            draw.rectangle((image_x+x*sx,image_y+y*sy,image_x+(x+w)*sx,image_y+(y+h)*sy),outline=(220,35,35),width=3)
            draw.text((image_x+x*sx,max(image_y,image_y+y*sy-15)),"bbox 누락 의심 위치",fill=(220,35,35),font=label_font,stroke_width=1,stroke_fill="white")
    display(sheet)
```
